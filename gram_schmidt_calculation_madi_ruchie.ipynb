# The orginal messy vectors

```python
r1 = np.array ([3,0,3,0])
r2= np.array([3,2,3,3])
r3 = np.array ([-7,-4,-5,-6])
r4 = np.array([-7,-3,-11,2])
```

```python
print(r1)
```

```python
#Process the first vector r1 and create a "nice" vector u1

#select a vector
t = r1

# subtract off how much this vector points in the directions of our other orthonormal vectors.
# since we don't have any orthonormal vectors, this first vector will skip this step
n = t

# find the size of n
np.sqrt(n.dot(n))

#normalize the vector by dividing it by it's size. This will create our first nice vector
u1 = n/np.sqrt(n.dot(n))
print(u1)
```

```python
# Let's check our nice vector is normal and orthogonal to the other nice vectors
print("The size of our new vector is", u1.dot(u1))

print("Is the vector orthogonal to our existing u's?", "We don't have any other u's at this step.")
```

```python
#Process the second vector r2 and create a "nice" vector u2

#select a vector
t = r2

# subtract off how much this vector points in the directions of our other orthonormal vectors.
n = t - t.dot(u1)*u1

# find the size of n
np.sqrt(n.dot(n))

#normalize the vector by dividing it by it's size. This will create our second nice vector
u2 = n/np.sqrt(n.dot(n))
print(u2)
```

```python
# Let's check our nice vector is normal and orthogonal to the other nice vectors
print("The size of our new vector is", u2.dot(u2))

print("Is the vector orthogonal to our existing u's?")
print(u2.dot(u1))
```

```python
#Process the third vector r3 and create a "nice" vector u3

#select a vector
t = r3

# subtract off how much this vector points in the directions of our other orthonormal vectors.
n = t - t.dot(u1)*u1 - t.dot(u2)*(u2)

# find the size of n
np.sqrt(n.dot(n))

#normalize the vector by dividing it by it's size. This will create our second nice vector
u3 = n/np.sqrt(n.dot(n))
print(u3)
```

```python
# Let's check our nice vector is normal and orthogonal to the other nice vectors
print("The size of our new vector is", u3.dot(u3))

print("Is the vector orthogonal to our existing u's?")
print(u3.dot(u1), "and", u3.dot(u2))
```

```python
#Process the last vector r4 and create a "nice" vector u4

#select a vector
t = r4

# subtract off how much this vector points in the directions of our other orthonormal vectors.
n = t - t.dot(u1)*u1 - t.dot(u2)*(u2) - t.dot(u3)*u3

# find the size of n
np.sqrt(n.dot(n))

#normalize the vector by dividing it by it's size. This will create our second nice vector
u4 = n/np.sqrt(n.dot(n))
print(u4)
```

```python
# Let's check our nice vector is normal and orthogonal to the other nice vectors
print("The size of our new vector is", u4.dot(u4))

print("Is the vector orthogonal to our existing u's?")
print(u4.dot(u1), "and", u4.dot(u2), "and", u4.dot(u3))
```
